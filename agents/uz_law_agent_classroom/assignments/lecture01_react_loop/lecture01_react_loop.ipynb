## A recorded run (read it from the file; this works before you write any code)

```python
run = json.loads((config.TRACES / "q2_under18.json").read_text(encoding="utf-8"))
print("Q:", run["question"])
for turn in run["trace"]:
    print(f"--- step {turn['step']}")
    print(turn["model"])
    print("Observation:", (turn["observation"] or "")[:200])
print("ANSWER:", run["answer"])
```

## Run it through your loop (needs Q1-Q4)

```python
answer, trace = agent.run(run["question"], Model("replay", "q2_under18"), max_steps=run["max_steps"])
assert trace == run["trace"], "your loop does not reproduce the recorded run"
print("\nsame trace:", trace == run["trace"])
```
